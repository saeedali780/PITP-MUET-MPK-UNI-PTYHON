# Topic 1 – JSON Module

1. Write a program to store details of 3 books (title, author, price) into a JSON file.

```python
import json

# Step 1: Create a list of dictionaries (3 books ka data)
books = [
    {"title": "Python Basics", "author": "John Smith", "price": 450},
    {"title": "Learn JavaScript", "author": "Ali Khan", "price": 600},
    {"title": "Data Science Guide", "author": "Sara Ahmed", "price": 750}
]

# Step 2: Write this data to a JSON file
with open("books.json", "w") as file:
    json.dump(books, file, indent=4)  # indent=4 se file readable banegi

print("✅ Books data has been saved to 'books.json'")
```

2. Read the JSON file and print only the book titles. 


```python
import json

with open("books.json", "r") as file:
    books = json.load(file)
for book in books:
    print(f"Title: {book['title']}")
```
