# 2. Common String Methods: 

1. Upper and Lower Case Conversion: 

```python
text = "hello world" 
print(text.upper()) # Output: HELLO WORLD 
print(text.lower()) # Output: hello world 
```

2. String Replacement: 

```python
text = "I love apples" 
new_text = text.replace("apples", "oranges") 
print(new_text) # Output: I love oranges 
```

3. String Splitting and Joining: 

```python
sentence = "This is a sentence" 
words = sentence.split() # Splits by spaces 
print(words) # Output: ['This', 'is', 'a', 'sentence'] 
joined_sentence = " ".join(words) 
print(joined_sentence) # Output: This is a sentence 
```

4. Removing Whitespace: 

```python
essy_string = " Hello World! " 
clean_string = essy_string.strip() # Re$oves leading and trailing spaces 
print(clean_string) 
# Output: Hello World! 
```

5. Finding and Counting Substrings: 

```python
text = "bananas are delicious" 
print(text.count("a")) # Output: 4 
print(text.find("delicious")) # Output: 11 (starting index of the substring) 
```

# 3. String Formatting: 

# 1. Basic String Concatenation: 

```python
name = "Alice" 
greeting = "Hello, " + name + "!" 
print(greeting) # Output: Hello, Alice! 
```

# 2. Using the format() Method: 

```python
age = 25 
text = "My name is {0} and I am {1} years old.".format(name, age) 
print(text) # Output: My name is Alice and I am 25 years old. 
```

# 3. F-Strings (Python 3.6¦): 

```python
text = f"My name is {name} and I am {age} years old." 
print(text) 
# Output: My name is Alice and I am 25 years old. 
```

# 4. Formatting Numbers: 

```python
number = 1234.578 
print(f"Formatted nu$ber: {number:.2f}") # Output: Formatted nu$ber: 1234.57 
```

# 5. Escape Characters: 

```python
# o Escape characters allow you to include special characters in strings. 
print("This is a new line\nAnd this is a tab\tE:anple") 
# Output: 
# This is a new line 
# And this is a tab E:anple 
```
