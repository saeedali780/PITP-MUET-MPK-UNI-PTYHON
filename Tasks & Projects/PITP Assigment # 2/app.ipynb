# 1. INSTANCE METHOD

Q:1 create a class car having an attribute model and brand. add a method  show_info to print values

```python
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year
    def display_info(self):
        return f"{self.year} {self.brand} {self.model}"

car1 = Car("Toyota", "Corolla", 2020)
print(car1.display_info())
```

Q:2 create a class book having an attribute author title add method detail()

```python
class Book:
    def __init__(self, author, title):
        self.author = author
        self.title = title
    def detail(self):
        return f"'{self.title}' by {self.author}"

book1 = Book("John Smith", "How Nations Fail")
book1.detail()
```

# 2. CLASS METHOD

Q:1 create a class company with class variable company_name make a class method change_name() to change the name of company

```python
class Company:
    def __init__(self, name, location):
        self.name = name
        self.location = location
    def change_company_name(self, new_name):
        self.name = new_name
company1 = Company("TechCorp", "New York")
company1.change_company_name("InnoTech")
print(company1.name)
```

Q2: create a class of laptop with class variable company_name make a class method set_brand() to change the brand name

```python
class Laptop:
    def __init__(self, brand, processor, ram):
        self.brand = brand
        self.processor = processor
        self.ram = ram
    def upgrade_ram(self, additional_ram):
        self.ram += additional_ram
    def set_brand(self, new_brand):
        self.brand = new_brand

laptop1 = Laptop("Dell", "Intel i5", 8)
laptop1.upgrade_ram(8)
laptop1.set_brand("HP")
print(f"{laptop1.brand} laptop with {laptop1.ram}GB RAM")
```

# 3. STATIC METHOD

Q1: create a class of calculator and create a static method to subtract(a, b) and return subtraction of two numbers

```python
class Calculator:
    @staticmethod
    def subtract(a, b):
        return a - b

result = Calculator.subtract(10, 4)
print(f"Subtraction Result: {result}")
```

Q2:create a class of percentage with a static method give a value total no and obtained no and return percentage

```python
class Percentage:
    @staticmethod
    def calculate_percentage(obtained, total):
        if total == 0:
            return 0
        return (obtained / total) * 100
percent = Percentage.calculate_percentage(45, 50)
print(f"Percentage: {percent}%")
```
